```python
def calculate_zakat():
  savings=int(input("Enter your savings amount in pkr:"))
  zakat=savings*(2.5/100)
  print("your zakat amount is", zakat)

calculate_zakat()

```

```python

def calculate_bill():
  units=int(input("enter amount of units"))
  rate=int(input("enter the rate per unit"))
  elecbill=units*rate
  print("Electricity bill:", elecbill, "PKR")

calculate_bill()
```

```python
def calculate_profit():
  cost=int(input("Enter the cost price:"))
  sell=int(input("Enter the sell price:"))
  profit=sell-cost
  print("Profit:", profit, "PKR")

calculate_profit()
```
